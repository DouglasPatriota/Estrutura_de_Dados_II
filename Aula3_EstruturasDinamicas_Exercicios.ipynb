```
# ==============================================================================
# PARTE 1: A Ilusão da Memória, Referências e Mutabilidade em Python
# ==============================================================================

def demonstrar_referencias():
    # Imprime um cabeçalho para a seção de Identidade de Objetos.
    print("--- 1. Identidade de Objetos (id) ---")

    # Cria um paciente 'p1'.
    p1 = Paciente("João", 30)
    # Cria uma variável 'p2' e atribui o valor de 'p1'.
    # Neste ponto, 'p2' refere-se ao *mesmo* objeto que 'p1' na memória, pois não foi criado um novo Paciente, apenas uma nova referência para o mesmo objeto.
    p2 = p1

    # A função id() retorna o identificador único de um objeto, que geralmente corresponde ao seu endereço de memória.
    # Mostra o paciente 'p1' e seu ID.
    print(f"Paciente p1: {p1} | id(p1): {id(p1)}")
    # Mostra o paciente 'p2' e seu ID. Observará que o ID de 'p1' e 'p2' são os mesmos.
    print(f"Paciente p2: {p2} | id(p2): {id(p2)}")
    # Confirma se ambos os IDs são idênticos, provando que 'p1' e 'p2' referem-se ao mesmo objeto.
    print(f"p1 e p2 apontam para o mesmo objeto? {id(p1) == id(p2)}\n")

    # Imprime um cabeçalho para a seção de Mutabilidade vs Imutabilidade.
    print("--- 2. Mutabilidade vs Imutabilidade ---")

    # --- Demonstração com Tipo Imutável (Inteiro - idade do paciente) ---
    # A idade do paciente é um inteiro. Inteiros são tipos imutáveis.
    idade = p1.idade
    # Armazena o ID original da idade.
    id_idade_antes = id(idade)
    # Altera a idade somando 1. Como inteiros são imutáveis, esta operação
    # não modifica o objeto original da idade. Em vez disso, um *novo* objeto inteiro é criado
    # na memória, e 'idade' passa a referenciar este novo objeto.
    idade = idade + 1
    # Compara o ID da idade antes e depois da modificação. Eles serão diferentes.
    print(f"idade antiga id: {id_idade_antes} -> idade nova id: {id(idade)} (Objeto foi recriado)")

    # --- Demonstração com Tipo Mutável (Lista - condições do paciente) ---
    # 'condicoes1' recebe uma lista de condições de saúde. Listas são tipos mutáveis.
    condicoes1 = ["diabetes"]
    # Armazena o ID original de 'condicoes1'.
    id_condicoes1_antes = id(condicoes1)
    # Adiciona uma nova condição à lista usando o método .append().
    # Como listas são mutáveis, esta operação modifica o *conteúdo* do objeto lista
    # diretamente na memória. Nenhum novo objeto lista é criado.
    condicoes1.append("obesidade")
    # Compara o ID de 'condicoes1' antes e depois da modificação. Eles permanecerão os mesmos.
    print(f"Condições id antes: {id_condicoes1_antes} -> Condições id depois: {id(condicoes1)} (Objeto alterado no mesmo local)\n")

    # Imprime um cabeçalho para a seção de Referência vs Cópia de Listas.
    print("--- 3. Referência vs Cópia de Listas ---")

    # --- Atribuição por Referência (Alias) ---
    # 'fila_original' recebe uma lista com os nomes dos pacientes na fila.
    fila_original = ["João", "Maria", "Vovó Ana"]
    # 'fila_referencia' recebe 'fila_original'. Isso não cria uma nova lista.
    # Ambas as variáveis agora apontam para o *mesmo objeto* lista na memória.
    fila_referencia = fila_original

    # --- Criação de uma Cópia Independente (Slice) ---
    # 'fila_copia' é criada usando um 'slice' completo ([:]).
    # Este método cria uma *nova* lista na memória que contém os mesmos elementos da 'fila_original'.
    # 'fila_copia' é um objeto independente.
    fila_copia = fila_original[:]

    # --- Alterando a Fila Original ---
    # Adiciona um novo paciente à 'fila_original'.
    # Como 'fila_referencia' aponta para o mesmo objeto, ela também refletirá essa alteração.
    # 'fila_copia', sendo independente, não será afetada.
    fila_original.append("Vovô Bento")

    # Imprime os resultados para cada fila, incluindo seus IDs.
    # 'fila_original' mostrará o novo paciente e terá um ID.
    print(f"Original:   {fila_original} | id: {id(fila_original)}")
    # 'fila_referencia' mostrará o novo paciente e terá o MESMO ID que 'fila_original'.
    print(f"Referência: {fila_referencia} | id: {id(fila_referencia)} (Sofreu alteração!)")
    # 'fila_copia' NÃO mostrará o novo paciente e terá um ID DIFERENTE dos outros.
    print(f"Cópia:      {fila_copia} | id: {id(fila_copia)} (Permaneceu isolada)")

# Chama a função para executar todas as demonstrações quando o script é rodado.
demonstrar_referencias()
```

🏥 Exercício Prático: Sistema de Gerenciamento de Clínica

## 18. Problema real — Sistema de atendimento de uma clínica

Uma clínica recebe pacientes durante o dia. Cada paciente possui:

- nome;
- idade;
- prioridade.

Agora iremos aplicar os conceitos de referências criando uma Estrutura Dinâmica (Lista Encadeada / Fila).  

Classe Paciente: Representa os dados guardados em cada Nó.  

Classe Node: Contém o dado e a referência (proximo) para o próximo nó.  

Classe FilaClinica: Gerencia os apontadores inicio e fim da fila.  

Desafio: Implementação da inserção com Prioridade (Idade ≥ 60 anos).  

```
# ==============================================================================
# PARTE 2: Construção Dinâmica - Nó e Fila de Atendimento da Clínica
# ==============================================================================


class Paciente:
    """Classe responsável por armazenar as informações de cada paciente."""

    CONDICOES_PRIORITARIAS = ["diabetes", "pressao alta", "obesidade", "cardiopatia"]

    def __init__(self, nome: str, idade: int, condicoes: list = None):
        self.nome = nome
        self.idade = idade
        self.condicoes = condicoes if condicoes else []

        # Define se é prioritário com base na idade ou em condições de saúde
        condicoes_normalizadas = [c.lower().strip() for c in self.condicoes]
        tem_condicao_prioritaria = any(
            condicao in condicoes_normalizadas for condicao in self.CONDICOES_PRIORITARIAS
        )
        self.eh_prioritario = idade >= 60 or tem_condicao_prioritaria

    def __repr__(self):
        status = "PRIORITÁRIO" if self.eh_prioritario else "NORMAL"
        condicoes_str = f" | Condições: {', '.join(self.condicoes)}" if self.condicoes else ""
        return f"[{self.nome}, {self.idade} anos - {status}{condicoes_str}]"


class Node:
    """A estrutura básica de um Nó dinâmico contendo o dado e a referência do próximo."""
    def __init__(self, paciente: Paciente):
        self.dado = paciente    # Armazena a instância do paciente
        self.proximo = None    # Âncora/Referência para o próximo nó na corrente


class FilaClinica:
    """
    Gerenciador da Fila de Pacientes.
    Manipula as referências 'inicio' e 'fim' para adicionar e remover de forma eficiente.
    """
    def __init__(self):
        self.inicio = None
        self.fim = None
        self._tamanho = 0

    def esta_vazia(self) -> bool:
        """Verifica se a fila possui pacientes."""
        return self.inicio is None

    def adicionar(self, nome: str, idade: int, condicoes: list = None):
        """
        Adiciona um paciente respeitando as regras de prioridade:
        - Prioritários (idade >= 60 OU com doenças como diabetes, pressão alta,
          obesidade ou cardiopatia) entram antes dos normais, respeitando a ordem de chegada.
        - Não-prioritários vão sempre para o final da fila.
        """
        novo_paciente = Paciente(nome, idade, condicoes)
        novo_no = Node(novo_paciente)

        # CASO 1: Fila está totalmente vazia
        if self.esta_vazia():
            self.inicio = novo_no
            self.fim = novo_no
            self._tamanho += 1
            print(f"-> Paciente {novo_paciente.nome} adicionado como o primeiro da fila.")
            return

        # CASO 2: Paciente é PRIORITÁRIO
        if novo_paciente.eh_prioritario:
            # Subcaso 2.1: O primeiro paciente da fila é Normal -> Novo nó entra no início
            if not self.inicio.dado.eh_prioritario:
                novo_no.proximo = self.inicio
                self.inicio = novo_no
            else:
                # Subcaso 2.2: Percorre a fila para encontrar o último prioritário
                atual = self.inicio
                while atual.proximo is not None and atual.proximo.dado.eh_prioritario:
                    atual = atual.proximo

                # Reorganiza os ponteiros/referências para encaixar o nó prioritário
                novo_no.proximo = atual.proximo
                atual.proximo = novo_no

                # Se foi inserido no final da fila, atualiza a referência do fim
                if novo_no.proximo is None:
                    self.fim = novo_no

        # CASO 3: Paciente NÃO PRIORITÁRIO (Insere sempre no final)
        else:
            self.fim.proximo = novo_no
            self.fim = novo_no

        self._tamanho += 1
        print(f"-> Paciente {novo_paciente.nome} ({novo_paciente.idade} anos) adicionado à fila.")

    def atender(self):
        """Remove e retorna o primeiro paciente da fila (atendimento)."""
        if self.esta_vazia():
            print("\n⚠️ A fila está vazia! Nenhum paciente para atender.")
            return None

        # O primeiro da fila é selecionado
        paciente_atendido = self.inicio.dado
        # Move o ponteiro 'inicio' para o próximo da fila
        self.inicio = self.inicio.proximo
        self._tamanho -= 1

        # Se a fila ficou vazia após o atendimento, o fim também vira None
        if self.inicio is None:
            self.fim = None

        print(f"\n✅ Atendendo: {paciente_atendido.nome}")
        return paciente_atendido

    def listar_espera(self):
        """Percorre a fila do início ao fim exibindo todos os pacientes."""
        print("\n--- 📋 FILA DE ESPERA ATUAL ---")
        if self.esta_vazia():
            print("Nenhum paciente aguardando.")
            print("-------------------------------\n")
            return

        atual = self.inicio
        posicao = 1

        # IMPORTANTE: A variável 'atual' caminha para frente a cada iteração (atual = atual.proximo)
        # para evitar o ERRO CRÍTICO de Laço Infinito!
        while atual is not None:
            print(f"{posicao}º -> {atual.dado}")
            atual = atual.proximo  # Avança para o próximo nó
            posicao += 1
        print(f"Total na fila: {self._tamanho}")
        print("-------------------------------\n")
```

```
# ==============================================================================
# PARTE 3: Execução e Testes do Sistema de Atendimento
# ==============================================================================

# Instanciando a fila da clínica
clinica = FilaClinica()

print("=== 1. CHEGADA DE PACIENTES NORMAIS E PRIORITÁRIOS ===")
clinica.adicionar("João", 30)                              # Normal
clinica.adicionar("Maria", 25)                              # Normal
clinica.adicionar("Vovó Ana", 72)                           # Prioritário (idade) - Deve ir para a frente da fila
clinica.adicionar("Pedro", 40)                               # Normal
clinica.adicionar("Vovô Bento", 80)                          # Prioritário (idade) - Deve ir atrás da Vovó Ana, mas antes dos normais
clinica.adicionar("Carla", 28, ["diabetes"])                 # Prioritário (doença) - Deve ir atrás do Vovô Bento, mas antes dos normais
clinica.adicionar("Roberto", 35, ["pressao alta", "obesidade"]) # Prioritário (doença) - Deve ir atrás da Carla, mas antes dos normais

# Exibe o estado da fila
clinica.listar_espera()

print("=== 2. ATENDIMENTO DOS PACIENTES ===")
# Atendendo o primeiro prioritário (Vovó Ana)
clinica.atender()
clinica.listar_espera()

# Atendendo o segundo prioritário (Vovô Bento)
clinica.atender()
clinica.listar_espera()

# Atendendo o terceiro prioritário (Carla - diabetes)
clinica.atender()
clinica.listar_espera()

# Atendendo o quarto prioritário (Roberto - pressão alta e obesidade)
clinica.atender()
clinica.listar_espera()

print("=== 3. ATENDENDO O RESTANTE DOS PACIENTES ===")
clinica.atender() # João
clinica.atender() # Maria
clinica.atender() # Pedro

# Tentando atender com fila vazia
clinica.atender()
clinica.listar_espera()
```
